```python
import sympy as sp
from sympy import symbols, Matrix, expand, factor, solve
sin = sp.sin
cos = sp.cos
l1, l2, l3 = symbols('l1 l2 l3')
fx, fy, fz = symbols('fx fy fz')
theta1, theta2, theta3 = symbols('theta1 theta2 theta3')
dtheta1, dtheta2, dtheta3 = symbols('dtheta1 dtheta2 dtheta3')
c1, s1, c2, s2, c3, s3 = cos(theta1), sin(theta1), cos(theta2), sin(theta2), cos(theta3), sin(theta3)
c12, s12 = c1*c2 - s1*s2, s1*c2 + c1*s2
c23, s23 = c2*c3 - s2*s3, s2*c3 + c2*s3
c13, s13 = c1*c3 - s1*s3, s1*c3 + c1*s3
c123, s123 = c12*c3 - s12*s3, s12*c3 + c12*s3

def show(expr):
    c1, s1, c2, s2, c3, s3 = symbols('c1 s1 c2 s2 c3 s3')
    c12, s12 = symbols('c12 s12')
    c23, s23 = symbols('c23 s23')
    c13, s13 = symbols('c13 s13')
    c123, s123 = symbols('c123 s123')

    replace = {sin(theta1): s1, cos(theta1): c1,
            sin(theta2): s2, cos(theta2): c2,
            sin(theta3): s3, cos(theta3): c3,
            sin(theta2 + theta3): s23, cos(theta2 + theta3): c23,
            sin(theta1 + theta2): s12, cos(theta1 + theta2): c12,
            sin(theta1 + theta3): s13, cos(theta1 + theta3): c13,
            sin(theta1 + theta2 + theta3): s123, cos(theta1 + theta2 + theta3): c123,
            }
    
    expr = expr.subs(replace)
    
    return expr
```

```python
m = Matrix([[sp.cos(theta1), -sp.sin(theta1)], [sp.sin(theta1), sp.cos(theta1)]])

# 输出逆矩阵
m_inv = m.inv()
m_inv_ = sp.simplify(m_inv)
sp.pprint(m_inv, use_unicode=True)
sp.pprint(m_inv_, use_unicode=True)

# 输出行列式
sp.pprint(m.det(), use_unicode=True)

# 输出特征值
sp.pprint(m.eigenvals(), use_unicode=True)

# 输出特征向量
sp.pprint(m.eigenvects(), use_unicode=True)

```

```python
# 输出向量积
v1 = Matrix([c1, s1, c3])
v2 = Matrix([c2, s2, c3])
res = v1.cross(v2)
sp.pprint(res, use_unicode=True)

# 输出向量积的模
sp.pprint(v1.cross(v2).norm(), use_unicode=True)
```

```python
# 计算theta微分
dtheta1, dtheta2, dtheta3 = sp.diff(theta1, 't'), sp.diff(theta2, 't'), sp.diff(theta3, 't')
sp.pprint(dtheta1, use_unicode=True)

# 展开
expr = (c1 + s1) * (c2 + s2) * (c3 + s3)
expanded_expr = expand(expr)
sp.pprint(expanded_expr, use_unicode=True)

# 因式分解
factored_expr = factor(expanded_expr)
sp.pprint(factored_expr, use_unicode=True)
```

```python
# 5.3
## 5.3（3）
m1 = Matrix([[-s1*(l1+l2*c2+l3*c23),-l2*c1*s2-l3*c1*s23, -l3*c1*s23],
             [c1*(l1+l2*c2+l3*c23), -l2*s1*s2-l3*s1*s23, -l3*s1*s23],
             [0, l2*c2+l3*c23, l3*c23]])
m2 = Matrix([[c1*c23,s1*c23,s23],
             [-c1*s23,-s1*s23,c23],
             [s1,-c1,0]])

res = m2*m1
# for i in res:
#     print(sp.simplify(i))

res = res.applyfunc(sp.simplify)
res = show(res)
sp.pprint(res, use_unicode=True)
```

```python
# 5.12
m1 = Matrix([[-s1*(l1+l2*c2+l3*c23),-l2*c1*s2-l3*c1*s23, -l3*c1*s23],
             [c1*(l1+l2*c2+l3*c23), -l2*s1*s2-l3*s1*s23, -l3*s1*s23],
             [0, l2*c2+l3*c23, l3*c23]])
m2 = Matrix([[c1*c23,s1*c23,s23],
             [-c1*s23,-s1*s23,c23],
             [s1,-c1,0]])

res = m2*m1
# for i in res:
#     print(sp.simplify(i))

res = res.applyfunc(sp.simplify)
res = res.det()
res = sp.factor(res)
res = show(res)

sp.pprint(res, use_unicode=True)
```
