# 二进制转换
二进制与十进制转换，使用字符串传入，支持二进制小数
能够将小数转换为ieee 754格式

```python
def bin2dec(binary_str):
    integer_part, fractional_part = binary_str.split('.')
    decimal_integer_part = int(integer_part, 2)
    decimal_fractional_part = sum(int(bit) * 2 ** (-i - 1) for i, bit in enumerate(fractional_part))
    return decimal_integer_part + decimal_fractional_part

def dec2bin(decimal_str, precision=20):
    # 处理整数部分和小数部分
    if '.' in decimal_str:
        integer_part_str, fractional_part_str = decimal_str.split('.')
    else:
        integer_part_str = decimal_str
        fractional_part_str = '0'
    
    # 转换整数部分
    integer_part = int(integer_part_str)
    binary_integer = bin(integer_part)[2:]
    
    # 转换小数部分
    fractional_part = float('0.' + fractional_part_str)
    binary_fractional = []
    
    for _ in range(precision):
        if fractional_part == 0:
            break
        fractional_part *= 2
        bit = int(fractional_part)
        binary_fractional.append(str(bit))
        fractional_part -= bit
    
    binary_fractional_str = ''.join(binary_fractional)
    
    # 组合整数部分和小数部分
    binary_str = binary_integer
    if binary_fractional_str:
        binary_str += '.' + binary_fractional_str
    
    return binary_str
```

```python
import struct

def float2ieee754(number, precision='single'):
    if precision == 'half':
        # 半精度（16位）
        packed = struct.pack('>f', number)
        unpacked = struct.unpack('>d', packed)[0]
        ieee754 = struct.pack('>e', unpacked)
    elif precision == 'single':
        # 单精度（32位）
        ieee754 = struct.pack('>f', number)
    elif precision == 'double':
        # 双精度（64位）
        ieee754 = struct.pack('>d', number)
    else:
        raise ValueError("Precision must be 'half', 'single', or 'double'")
    
    # 将二进制数据转换为十六进制字符串
    hex_str = ieee754.hex()
    return hex_str

print(float2ieee754(0.1, 'single'))
```
